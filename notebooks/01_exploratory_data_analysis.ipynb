### What this notebook does

**Exploratory Data Analysis (EDA)** is the first step of any machine learning
project: looking at the raw data before building any model, to understand
its shape, types, distributions, and quality (missing values, duplicates,
obvious outliers).

This notebook loads all three datasets used across the project
(`Social_Network_Ads.csv`, `Mall_Customers.csv`, `insurance_pre.csv`),
prints basic summary statistics for each, and renders one combined chart
showing their key distributions. No model is trained here — this notebook
only looks at the data.
